# Exercise 5.A NumPy Fundamentals
This notebook practices creating NumPy arrays, checking data types, transposing, indexing, and building multi-dimensional arrays.

```python
import numpy as np
```

```python
fruits = ['apple', 'banana', 'cherry', 'date']
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
arr_1.dtype
# Returns <U6: "U" means Unicode string (text),
# and "6" is the length of the longest string in the array ("banana" and "cherry" have 6 letters).
```

```python
arr_2 = np.array([(10, 11, 12), (13, 14, 15)])
print(arr_2)
```

```python
print(arr_2.T)
```

```python
print(arr_2)
# No, .T does not change arr_2. It only shows a transposed view; the original array stays the same.
```

```python
arr_3 = np.array([(1, 3, 5), (2, 4, 6)])
print(arr_3)
```

```python
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])
print(arr_4)
```

```python
arr_5 = np.ones((2, 2, 2))
print(arr_5)
```
