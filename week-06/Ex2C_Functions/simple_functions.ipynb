Defining and calling custom functions.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician is {music}."

print(favorite_things("Meric", "Shrek", "Hans Zimmer"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined Year Up United to turn my love of working with data into a career as a data analyst."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Antalya", "it is my home,food and the sea in one city"))
```

```python
print(favorite_place("Lake Tahoe", "the views are amazing and it's a great escape from the city"))
```
