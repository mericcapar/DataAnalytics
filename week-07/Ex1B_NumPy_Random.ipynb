```python
students = np.array(['Ali', 'Maria', 'James', 'Sofia', 'David', 'Emma',
                     'Lucas', 'Olivia', 'Daniel', 'Mia', 'Ethan', 'Ava'])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# Not effective: choice() picks with replacement by default, so the same student
# can show up in more than one group and some students may be left out.
# Using replace=False would fix this.
```

```python
p_values = [0.3, 0, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07, 0.07]
call_on_weighted = np.random.choice(students, p=p_values, size=5)
print(call_on_weighted)
# Ali has a 30% chance (0.3), Maria has 0% (0) so she is never picked,
# and the other 10 students share the remaining 70% equally (0.07 each).
# All p values must add up to 1.
```

## Generating a Normal Distribution
