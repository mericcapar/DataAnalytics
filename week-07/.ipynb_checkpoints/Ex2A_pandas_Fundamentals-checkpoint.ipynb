# Exercise 2.A Fundamentals of pandas

```python
import numpy as np
import pandas as pd
```

## Series 

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
try:
    ser3 = pd.Series(ac)
except TypeError as e:
    print("Error:", e)
    ser3 = pd.Series(sorted(ac))   # a set has no order, so turn it into a sorted list first
ser4 = pd.Series(ad)

print(ser1)
print(ser2)
print(ser3)
print(ser4)
```

```python
try:
    ser5 = pd.Series(17, 18, 19, 20)
except TypeError as e:
    print("Error:", e)
# TypeError: the values must be passed as one collection, like a list [17, 18, 19, 20].
# Passing them as separate arguments makes pandas read 18, 19 and 20 as other settings (like index).
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1)
df1
```

```python
df1.insert(1, 'ser2', ser2)
df1.insert(2, 'ser3', ser3)
df1
```

```python
df1.insert(3, 'ser4', ser4)
df1
# The ser4 column is all NaN. ser4 uses the dictionary keys as its index,
# but df1 uses 0, 1, 2, 3. pandas matches values by index label, finds no matches, and fills NaN.
```

```python
df1.loc[4] = [5, 9, 12, 0]
df1
```

## Reading Data into a DataFrame

```python
df2 = pd.read_csv("sakila_example.csv")
df2
```

```python
df2.head()
```

```python
df2.tail(10)
```

```python
df2.info()
```

The data has 262 rows and 5 columns: title, description, category, language and actor_name. Every column is text, and there are no missing values since all columns show 262 non-null entries. Each row is one actor in one film, so the same film appears on several rows.

```python
df2.describe()
```

Since every column is text, describe shows count, unique, top and freq instead of averages. Useful insights: there are 46 unique films, 145 unique actors and 15 categories. Horror is the most common category (55 rows), and Oprah Kilmer appears most often (8 rows). Less useful: language has only one value (English), and the top description just repeats the film with the most actors, so it doesn't tell us much.

```python
df3 = pd.read_csv("northwind_example.csv")
df3
```

```python
df3.describe()
```

This data has numeric columns, so describe gives real statistics: mean, standard deviation, min, max and quartiles. The average unit price is about $23.37, but the median is only $15.80 and the max is $210.80, so a few expensive products pull the average up. The average quantity per order line is about 24 items, ranging from 1 to 120.
