# Exercise 3.A Getting familiar with Matplotlib


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker ='o')
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
plt.plot(x,y, linestyle='--', color='red')
plt.show()
```

```python
plt.plot(x, y, marker='s', markersize=10, markerfacecolor='green')
plt.show()
```

```python
plt.plot(x, y, marker='D', markersize=10, markerfacecolor='green')
plt.show()
```

```python
plt.plot(x, y, marker='d', markersize=10, markerfacecolor='green')
plt.show()
```

```python
plt.plot(x, y, marker='^', markersize=10, color='purple')   # triangles
plt.show()
```

```python
plt.plot(x, y, marker='*', markersize=12, color='yellow')   # stars
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
plt.bar(products,sales)
plt.show()
```

```python
plt.bar(products,sales, color='skyblue', width=0.5)
plt.title('Sales by Product')
plt.xlabel('Product')
plt.ylabel('Units sold')
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y)
plt.show()
```

```python
sizes = np.random.rand(50) * 100
plt.scatter(x,y, s=sizes, c=sizes, cmap='viridis')
plt.colorbar()
plt.title('Random Scatter Plot')
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.show()
```

## Subplots

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Left: sine wave (line chart)
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax1.plot(x, y)
ax1.set_title('Sine Wave')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')

# Right: histogram of 1000 random numbers
data = np.random.randn(1000)
ax2.hist(data, bins=30)
ax2.set_title('Histogram of Random Numbers')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')

plt.show()
```
