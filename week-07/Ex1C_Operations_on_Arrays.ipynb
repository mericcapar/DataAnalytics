## Sample Arrays 

```python
r_1 = np.random.randint(1, 20, size=2)
r_2 = np.random.randint(1, 20, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 100, size=(4, 5))
r_5 = np.random.rand(4, 5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

Shuffle only changed the order of the rows. Each row kept the same values in the same order, and the rows just moved to new positions.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

shuffle() changes the original array in place, so I would use it when I don't need the original order anymore. permutation() returns a shuffled copy and leaves the original unchanged, so I would use it when I want to keep the original data.

## Mathematical Operations

```python
print(r_1, r_1.shape)
print(r_2, r_2.shape)
result = np.add(r_1, r_2)
print(result, result.shape)
```

```python
print(r_1, r_1.shape)
print(r_3, r_3.shape)
result = np.add(r_1, r_3)
print(result, result.shape)
```

```python
print(r_2, r_2.shape)
print(r_4, r_4.shape)
try:
    print(np.add(r_2, r_4))
except ValueError as e:
    print("Error:", e)
# Error: operands could not be broadcast together with shapes (1,2) (4,5).
# r_2 has 2 columns and r_4 has 5, so NumPy can't line them up.
```

```python
print(np.add(r_4, np.array([1])))
```

```python
print(np.add(r_4, np.array([[1]])))
```

```python
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))
```

```python
try:
    print(np.add(r_4, np.array([[1], [1]])))
except ValueError as e:
    print("Error:", e)
# Error: operands could not be broadcast together with shapes (4,5) (2,1).
# The small array has 2 rows but r_4 has 4, and 2 can't stretch to 4.
```

Broadcasting lets NumPy add two arrays of different shapes by stretching the smaller one to match the bigger one. NumPy compares the shapes from right to left, and each dimension must either be equal or be 1. A dimension of 1 gets copied to fit. The result takes the larger size in each dimension. If a dimension is neither equal nor 1, NumPy gives an error.

## Cumulative Operations

```python
print(r_4)
```

```python
print(np.sum(r_4))
```

```python
print(np.sum(r_4, axis=0))   # down the rows -> one total per column (5 values)
```

```python
print(np.sum(r_4, axis=1))   # across the columns -> one total per row (4 values)
```

```python
print(np.cumsum(r_4))        # running total of all values, flattened to 1D
```

```python
print(np.cumsum(r_4, axis=0))  # running total down each column
```

## Grocery Proble

```python
# order: vegetables, fruits, snacks, beverages
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print(bag1)
print(bag2)
```

```python
print(f"The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.")
```

```python
totals = np.add(bag1, bag2)
print(f"There are {totals[0]} vegetables, {totals[1]} fruits, {totals[2]} boxes of snacks, and {totals[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
ben = np.array([0, 5, 5, 3, 4])    # missed Monday -> 0 doors
```

```python
carla = np.array([3, 6, 0, 3, 4])  # sick on Wednesday -> 0 doors
```

```python
all_knocks = np.vstack((ali, ben, carla))
print(all_knocks)
```

```python
per_person = np.sum(all_knocks, axis=1)
print(f"Ali: {per_person[0]}, Ben: {per_person[1]}, Carla: {per_person[2]}")
print(f"Team total: {np.sum(all_knocks)}")
```

```python
daily_totals = np.sum(all_knocks, axis=0)   # doors per day
rolling_votes = np.cumsum(daily_totals) * 0.5
print(rolling_votes)
```
