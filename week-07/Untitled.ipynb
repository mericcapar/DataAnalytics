# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([(11,12,13),(14,15,16)])
print(arrA)
```

```python
for i in arrA:
    print(i*2)
```

```python
for i in arrA:
    for j in i:
        print(j*2)
```

### First one is for all the list but the second one lists them all indiviual

```python
for x in arrA:
    for y in x:
        loop_result = x *2
print(loop_result)
# loop_result doesn't keep the first row and it skips it. It only holds the second row. 
```

 ## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA,arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA,arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1= np.concatenate((joined_array_1,joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
```

```python
# joined_array_1 is 4x3 and joined_array_2 is 2x6, so they can't be stacked.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
```

```python
## To join side by side, both arrays need the same number of rows.
```

```python
h_stack = np.hstack((arrA, arrA))  
v_stack = np.vstack((arrA, arrA))  
d_stack = np.dstack((arrA, arrA))  

print(h_stack)
print(v_stack)
print(d_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# Result: 2 arrays, each 2-dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# Result: 3 arrays, each 2-dimensional. 4 rows can't split evenly into 3,
# so the first array gets 2 rows and the other two get 1 row each.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([(7, 6), (13, 5), (7, 3)])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
